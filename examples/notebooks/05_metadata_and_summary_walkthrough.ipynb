# Metadata and Summary Walkthrough

BioSeqDownloader workflow runs can write `metadata.json` and `run_summary.yml` when enabled in the descriptor.

This notebook validates a descriptor offline and inspects existing outputs if they are already present. It does not require credentials.

```python
import json
from pathlib import Path
from pprint import pprint

import yaml

from bioseq_dl.cli.workflows import load_workflow_recipe, validate_workflow_recipe

repo_root = Path.cwd().parent.parent
config_path = repo_root / "examples" / "workflows" / "protein_query_first_minimal.yml"
output_dir = repo_root / "examples" / "results" / "protein_query_first_minimal"

recipe = load_workflow_recipe(config_path)
normalized = validate_workflow_recipe(recipe)

pprint({
    "schema_version": normalized["schema_version"],
    "metadata_enabled": normalized["include_metadata"],
    "summary_enabled": normalized["include_summary"],
    "output": normalized["output"],
})
```

## Expected metadata structure

After execution, `metadata.json` is expected to include `schema_version` inside descriptors plus these top-level keys:

- `tool`
- `original_descriptor`
- `normalized_descriptor`
- `normalized_workflow_values`
- `execution`
- `output_files`
- `reporting`

```python
metadata_path = output_dir / "metadata.json"
summary_path = output_dir / "run_summary.yml"

if metadata_path.exists():
    metadata = json.loads(metadata_path.read_text(encoding="utf-8"))
    pprint({
        "metadata_keys": sorted(metadata),
        "tool": metadata.get("tool"),
        "original_schema_version": metadata.get("original_descriptor", {}).get("schema_version"),
        "normalized_schema_version": metadata.get("normalized_descriptor", {}).get("schema_version"),
    })
else:
    print("metadata.json does not exist yet. It is produced after a workflow run when include_metadata is true.")

if summary_path.exists():
    summary = yaml.safe_load(summary_path.read_text(encoding="utf-8"))
    pprint({
        "summary_keys": sorted(summary),
        "tool": summary.get("tool"),
        "schema_version": summary.get("schema_version"),
    })
else:
    print("run_summary.yml does not exist yet. It is produced after a workflow run when include_summary is true.")
```
