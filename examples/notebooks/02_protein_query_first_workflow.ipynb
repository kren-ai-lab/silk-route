# Protein Query-First Workflow

`query_first` runs one executable `query.value` string for a selected modality. This notebook uses the minimal protein workflow descriptor.

Running the workflow calls live APIs and requires internet access. If the API or network is unavailable, the command may fail; the notebook handles that case gracefully.

```python
import json
import subprocess
import sys
from pathlib import Path

repo_root = Path.cwd().parent.parent
config_path = repo_root / "examples" / "workflows" / "protein_query_first_minimal.yml"
output_dir = repo_root / "examples" / "results" / "protein_query_first_minimal"

print(config_path)
```

## Run the workflow

The command below uses the module form so it runs against the current Python environment.

```python
command = [
    sys.executable,
    "-m",
    "silkroute.cli.main",
    "workflow",
    "run",
    "--config",
    str(config_path),
]

result = subprocess.run(command, cwd=repo_root, capture_output=True, text=True, check=False)
print(result.stdout)

if result.returncode != 0:
    print("Workflow execution did not complete. This can happen when internet access or an API service is unavailable.")
    print(result.stderr)
else:
    print("Workflow completed.")
```

## Inspect output files

When execution succeeds, the workflow writes result files plus metadata and a run summary under the configured output directory.

```python
if output_dir.exists():
    print("Output files:")
    for path in sorted(output_dir.iterdir()):
        print("-", path.relative_to(repo_root))

    metadata_path = output_dir / "metadata.json"
    if metadata_path.exists():
        metadata = json.loads(metadata_path.read_text(encoding="utf-8"))
        print("Metadata keys:", sorted(metadata))
else:
    print("No output directory exists yet. Run the workflow when internet access is available.")
```
