# BioSeqDownloader API Usage Demo
This document demonstrates how to use the BioSeqDownloader API implementations programmatically.

### Importing the API
To use the BioSeqDownloader API, you first need to import the necessary classes from the `BioSeqDownloader` package.

Every API implementation is located in its own module inside the `bioseq_dl` package. For every database available, there is a corresponding interface class that provides methods to interact with that database. We use the `{database_name}Interface` naming convention for these classes.

```python
from bioseq_dl import AlphafoldInterface
```

### Instantiating the API
Different API have different requirements for instantiation. Some may require an email address, while others may need an API key or token. Please refer to the specific database documentation for details on how to instantiate the API correctly.

```python
instance = AlphafoldInterface(
    structures=["pdb"],
)
```

### Making a Request
Once you have instantiated the API, you can use its methods to make requests to the database. Every API class provides 2 main methods for making requests:
- `fetch_single`: This method is used to fetch data for a single identifier.
- `fetch_batch`: This method is used to fetch data for multiple identifiers in a batch.

In some databases `fetch_single` can be used to fetch data for multiple identifiers as well.

Every database has their own query definition and accepted parameters. Please refer to the specific database documentation for details on how to construct your queries.
Some common parameters include:
- `query`: The identifier or list of identifiers to search for.
- `method`: It refers to a endpoint or a specific type of search.
- `parse`: A boolean indicating whether to parse the response into a structured format. By default it will use the configuration defined at instantiation. You can use your own fields by giving fields_to_extract parameter.
- `to_dataframe`: A boolean indicating whether to return the results as a pandas DataFrame. It's highly enccouraged to use this parameter with parse=True.

This will return 2 objects: the fetched data and metadata about the request. The fetched data can be a list, dictionary, or pandas DataFrame depending on the parameters used. The metadata contains additional information about the request, such as cached IDs.

```python
response, metadata = instance.fetch_single(
    query="P02666",
    method="prediction",
    parse=True,
    to_dataframe=True,
)
```

```python
response
```

```python
metadata
```

```python
response, metadata = instance.fetch_batch(
    queries=["P02666", "Q9TSI0", "P33048", "P11839", "O15552", "P76011"],
    method="prediction",
    parse=False,
    to_dataframe=True,
)
```

## Chembl

### Importing the API

```python
from bioseq_dl import ChEMBLInterface
```

### Instantiating the API

```python
instance = ChEMBLInterface()
```

### Making a Request

A number of methods are available to make requests to the database.

- `target`: Gather protein and non-protein targets information.
- `assay`: Assay details as reported in ChEMBL.
- `cell_line`: Cell line information.
- `molecule`: Molecule information, including properties, structural representations and synonyms.
- `activity`: Activity values recorded in an Assay.
- `binding_site`: Target binding site definition
- `substructure`: Substructure search for molecules.
- `similarity`: Similarity search for molecules.

For `target`, `assay`, `cell_line` and `molecule` you can define a query with either 'query' or 'filters' parameters. 

query: str or list of str
    Single identifier or list of identifiers to search for.

filters: dict
    Dictionary of filters to apply to the search. The keys are: `field`, `filter_type`, `value`.
More information of what fields and filter types are available in the [ChEMBL API documentation](https://chembl.gitbook.io/chembl-interface-documentation/web-services/chembl-data-web-services#filtering-and-ordering).

### Important
Additionaly for this API you can set the number of pages to fetch with `pages_to_fetch` parameter. By default it will fetch only the first page. Use -1 to fetch all available pages.

Using Target

```python
response, _ = instance.fetch_single(
    query={
        "filters": [
            {
                "field": "gene_symbol",
                "filter_type": "icontains",
                "value": "EGFR"
            }
        ]
    },
    method="target",
    parse=True,
    to_dataframe=True
)
response.head(5)
```

```python
response, _ = instance.fetch_single(
    query={
        "filters": [{
                "field": "molecule_properties__mw_freebase",
                "filter_type": "lte",
                "value": "300"
            },
            {
                "field": "pref_name",
                "filter_type": "iendswith",
                "value": "nib"
            }
        ]
    },
    method="molecule",
    parse=True,
    to_dataframe=True
)
response.head(5)
```
